```python
# Leemos el dato limpio que ya guardamos en Silver (no hace falta reprocesar todo)
df = spark.read.format('parquet')\
    .load('abfss://silver@covid19arturodatalake.dfs.core.windows.net/covid19_mexico_2022_limpio')

print(f"Registros cargados desde Silver: {df.count():,}")
```

```python
# ═══════════════════════════════════════════════════════
# CAPA GOLD — Datos finales listos para consumo analítico (Power BI)
# ═══════════════════════════════════════════════════════

df.write.format('parquet')\
    .mode('overwrite')\
    .option('path', 'abfss://gold@covid19arturodatalake.dfs.core.windows.net/covid19_mexico_2022_gold')\
    .save()

print("Datos Gold guardados correctamente para el año 2022")
```

```python
# Registrar nuestra tabla en la capa Gold de nuestro catalogo que creamos 
spark.sql("""
CREATE TABLE IF NOT EXISTS COVID19_CATALOG.gold.covid19_mexico_2022_gold
USING PARQUET
LOCATION 'abfss://gold@covid19arturodatalake.dfs.core.windows.net/covid19_mexico_2022_gold'
""")

print("Tabla Gold registrada en Unity Catalog")
```
