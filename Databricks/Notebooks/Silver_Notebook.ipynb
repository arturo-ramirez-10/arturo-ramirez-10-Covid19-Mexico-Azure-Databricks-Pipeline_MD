# DATA TRANSFORMATION

```python
# Imports y funciones auxiliares
from pyspark.sql import functions as F
from pyspark.sql.types import *
import unicodedata
import time
from datetime import datetime

# Función para quitar acentos (UDF)
def quitar_acentos(texto):
    if texto is None:
        return texto
    return ''.join(
        c for c in unicodedata.normalize('NFD', str(texto))
        if unicodedata.category(c) != 'Mn'
    )

quitar_acentos_udf = F.udf(quitar_acentos, StringType())
```

```python
# Metadatos de la corrida
ANO = "2022"
CONDICION = "B"
REPETICION = 3

timestamp_inicio_dt = datetime.now()
timestamp_inicio = timestamp_inicio_dt.strftime("%Y-%m-%d %H:%M:%S")
tiempo_inicio_seg = time.time()

print(f"Inicio corrida: {timestamp_inicio}")
print(f"Año: {ANO} | Condición: {CONDICION} | Repetición: {REPETICION}")
```

```python
# Corrección de acentos y fechas
print("Cargando archivo original 2022...")
total_original = df.count()
print(f"  Registros originales: {total_original:,}")

# Corrección de acentos
print("  Corrigiendo acentos...")
for campo in ["PAIS_NACIONALIDAD", "PAIS_ORIGEN"]:
    if campo in df.columns:
        df = df.withColumn(campo, F.trim(quitar_acentos_udf(F.col(campo))))

# Limpieza FECHA_DEF (9999-99-99 → null)
print("  Limpiando FECHA_DEF (9999-99-99 → vacio)...")
df = df.withColumn(
    "FECHA_DEF",
    F.when(F.col("FECHA_DEF") == "9999-99-99", None).otherwise(F.col("FECHA_DEF"))
)
df = df.withColumn("FECHA_DEF", F.to_date("FECHA_DEF"))

# Conversión de fechas
df = df.withColumn("FECHA_SINTOMAS", F.to_date("FECHA_SINTOMAS"))
df = df.withColumn("FECHA_ACTUALIZACION", F.to_date("FECHA_ACTUALIZACION"))
df = df.withColumn("FECHA_INGRESO", F.to_date("FECHA_INGRESO"))
```

```python
# Criterio de admisibilidad
campos_criticos = [
    "ID_REGISTRO", "FECHA_INGRESO", "SEXO", "EDAD", "ENTIDAD_UM",
    "SECTOR", "TIPO_PACIENTE", "FECHA_SINTOMAS",
    "CLASIFICACION_FINAL", "RESULTADO_ANTIGENO"
]
campos_presentes = [c for c in campos_criticos if c in df.columns]

# Completitud: promedio de % no-nulos entre los campos críticos
completitudes = []
for campo in campos_presentes:
    pct = df.filter(F.col(campo).isNotNull()).count() / total_original * 100
    completitudes.append(pct)
completitud_admision = sum(completitudes) / len(completitudes)

duplicados_brutos = total_original - df.dropDuplicates(["ID_REGISTRO"]).count()
pct_duplicados_brut = duplicados_brutos / total_original * 100

print(f"\n  [ADMISIBILIDAD]")
print(f"  Completitud campos críticos: {completitud_admision:.1f}% (umbral ≥80%)")
print(f"  Duplicados brutos:           {pct_duplicados_brut:.1f}% (umbral <20%)")

if completitud_admision < 80 or pct_duplicados_brut >= 20:
    raise ValueError(
        f"Archivo NO admisible: completitud={completitud_admision:.1f}%, "
        f"duplicados_brutos={pct_duplicados_brut:.1f}%"
    )
print("  → Archivo ADMITIDO para procesamiento formal\n")
```

```python
# Las 9 reglas de depuración
# 1. FECHA_SINTOMAS no nula
antes = df.count()
df = df.filter(F.col("FECHA_SINTOMAS").isNotNull())
excl_fecha_sintomas_nula = antes - df.count()
print(f"  [-] FECHA_SINTOMAS nula:        {excl_fecha_sintomas_nula:,}")

# 2. FECHA_ACTUALIZACION no nula
antes = df.count()
df = df.filter(F.col("FECHA_ACTUALIZACION").isNotNull())
excl_fecha_act_nula = antes - df.count()
print(f"  [-] FECHA_ACTUALIZACION nula:   {excl_fecha_act_nula:,}")

# 3. SECTOR válido: SSA=12, IMSS=4, ISSSTE=6
antes = df.count()
df = df.filter(F.col("SECTOR").isin([4, 6, 12]))
excl_sector_invalido = antes - df.count()
print(f"  [-] SECTOR invalido:            {excl_sector_invalido:,}")

# 4. CLASIFICACION_FINAL = 1 o 3
antes = df.count()
df = df.filter(F.col("CLASIFICACION_FINAL").isin([1, 3]))
excl_clasificacion = antes - df.count()
print(f"  [-] CLASIFICACION_FINAL != 1,3: {excl_clasificacion:,}")

# 5. Periodo: 01-ene-2022 al 31-dic-2022
FECHA_INICIO = "2022-01-01"
FECHA_FIN = "2022-12-31"
antes = df.count()
df = df.filter(
    (F.col("FECHA_SINTOMAS") >= FECHA_INICIO) &
    (F.col("FECHA_SINTOMAS") <= FECHA_FIN)
)
excl_fuera_periodo = antes - df.count()
print(f"  [-] Fuera del periodo 2022:     {excl_fuera_periodo:,}")

# 6. Calcular latencia (días y minutos)
df = df.withColumn(
    "latencia_dias",
    F.datediff(F.col("FECHA_ACTUALIZACION"), F.col("FECHA_SINTOMAS"))
)
df = df.withColumn("latencia_minutos", F.col("latencia_dias") * 24 * 60)

# 7. Latencia negativa
antes = df.count()
df = df.filter(F.col("latencia_dias") >= 0)
excl_latencia_negativa = antes - df.count()
print(f"  [-] Latencia negativa:          {excl_latencia_negativa:,}")

# 8. Latencia extrema > 1095 días
antes = df.count()
df = df.filter(F.col("latencia_dias") <= 1095)
excl_latencia_extrema = antes - df.count()
print(f"  [-] Latencia > 1095 dias:       {excl_latencia_extrema:,}")

# 9. Deduplicación por llave lógica (quedándonos con el más completo)
from pyspark.sql.window import Window

llave = ["ID_REGISTRO", "FECHA_SINTOMAS", "ENTIDAD_UM", "SECTOR"]
antes = df.count()

# Completitud por fila = cantidad de columnas no nulas
df = df.withColumn(
    "_completitud",
    sum(F.when(F.col(c).isNotNull(), 1).otherwise(0) for c in df.columns)
)

window_spec = Window.partitionBy(*llave).orderBy(F.col("_completitud").desc())
df = df.withColumn("_rank", F.row_number().over(window_spec))
df = df.filter(F.col("_rank") == 1).drop("_rank", "_completitud")

excl_duplicados = antes - df.count()
print(f"  [-] Duplicados eliminados:      {excl_duplicados:,}")
```

```python
# Timestamps finales y throughput
tiempo_fin_seg = time.time()
timestamp_fin_dt = datetime.now()
timestamp_fin = timestamp_fin_dt.strftime("%Y-%m-%d %H:%M:%S")
duracion_min = (tiempo_fin_seg - tiempo_inicio_seg) / 60

print(f"\n  Fin corrida: {timestamp_fin}")
print(f"  Duración:    {duracion_min:.2f} minutos")

total_final = df.count()
total_excluidos = total_original - total_final
throughput_reg_min = total_final / duracion_min if duracion_min > 0 else 0
```

```python
# Métricas de calidad post-depuración
campos_requeridos = [
    "ID_REGISTRO", "FECHA_INGRESO", "SEXO", "EDAD", "ENTIDAD_UM",
    "SECTOR", "TIPO_PACIENTE", "FECHA_SINTOMAS", "CLASIFICACION_FINAL",
    "RESULTADO_ANTIGENO"
]
campos_ok = [c for c in campos_requeridos if c in df.columns]

completitudes_final = []
for campo in campos_ok:
    pct = df.filter(F.col(campo).isNotNull()).count() / total_final * 100
    completitudes_final.append(pct)
completitud_pct = sum(completitudes_final) / len(completitudes_final)

duplicados_pct = (total_final - df.dropDuplicates(llave).count()) / total_final * 100

consistencia_pct = df.filter(
    F.col("FECHA_ACTUALIZACION") >= F.col("FECHA_SINTOMAS")
).count() / total_final * 100

print(f"\n  [CALIDAD POST-DEPURACIÓN]")
print(f"  Completitud:   {completitud_pct:.1f}% (umbral >95%)")
print(f"  Duplicados:    {duplicados_pct:.1f}% (umbral <5%)")
print(f"  Consistencia:  {consistencia_pct:.1f}% (umbral >90%)")
print(f"  Throughput:    {throughput_reg_min:,.0f} registros/min")
```

```python
# Conformidad estructural y por sector
CAMPOS_CATALOGO_SSA = 39
campos_en_archivo = len(df.columns)
conformidad_estructural = min(campos_en_archivo, CAMPOS_CATALOGO_SSA) / CAMPOS_CATALOGO_SSA * 100

print(f"\n  [CONFORMIDAD ESTRUCTURAL]")
print(f"  Campos en archivo: {campos_en_archivo} / {CAMPOS_CATALOGO_SSA} del catálogo")
print(f"  Conformidad estructural: {conformidad_estructural:.1f}%")

sector_nombres = {4: "IMSS", 6: "ISSSTE", 12: "SSA"}

conf_por_sector = df.groupBy("SECTOR").count().withColumnRenamed("count", "registros")
conf_por_sector = conf_por_sector.withColumn(
    "pct_del_total", F.round(F.col("registros") / total_final * 100, 1)
)

print(f"\n  [CONFORMIDAD DE REGISTROS POR SECTOR]")
print(f"  Total conformes: {total_final:,} / {total_final:,} = 100%")
display(conf_por_sector)
```

# DATA WRITING

```python
# ═══════════════════════════════════════════════════════
# CELDA FINAL — Guardar datos limpios + bitácora + conformidad por sector
# Antes de correr, solo cambia REPETICION (1, 2 o 3) arriba en tu celda de metadatos
# ═══════════════════════════════════════════════════════

from pyspark.sql import Row

# ── 1. Datos limpios (overwrite — solo necesitas 1 copia final, es igual en las 3 repeticiones) ──
df.write.format('parquet')\
    .mode('overwrite')\
    .option('path', 'abfss://silver@covid19arturodatalake.dfs.core.windows.net/covid19_mexico_2022_limpio')\
    .save()

# ── 2. Bitácora de esta corrida (append — se acumulan las 3 repeticiones) ──
bitacora_row = spark.createDataFrame([Row(
    Ano=ANO,
    Condicion=CONDICION,
    Repeticion=REPETICION,
    timestamp_inicio=timestamp_inicio,
    timestamp_fin=timestamp_fin,
    duracion_min=round(duracion_min, 4),
    throughput_reg_min=round(throughput_reg_min, 0),
    Registros_originales=total_original,
    Excl_FECHA_SINTOMAS_nula=excl_fecha_sintomas_nula,
    Excl_FECHA_ACT_nula=excl_fecha_act_nula,
    Excl_SECTOR_invalido=excl_sector_invalido,
    Excl_CLASIFICACION_FINAL=excl_clasificacion,
    Excl_fuera_periodo=excl_fuera_periodo,
    Excl_latencia_negativa=excl_latencia_negativa,
    Excl_latencia_extrema=excl_latencia_extrema,
    Excl_duplicados=excl_duplicados,
    Registros_finales=total_final,
    Porcentaje_retenido=round(total_final/total_original*100, 1),
    completitud_pct=round(completitud_pct, 1),
    duplicados_pct=round(duplicados_pct, 1),
    consistencia_pct=round(consistencia_pct, 1),
    conformidad_estructural_pct=round(conformidad_estructural, 1)
)])

bitacora_row.write.format('parquet')\
    .mode('append')\
    .option('path', 'abfss://silver@covid19arturodatalake.dfs.core.windows.net/bitacora_corridas')\
    .save()

# ── 3. Conformidad por sector de esta corrida (append — igual se acumulan las 3) ──
conf_por_sector_final = conf_por_sector.withColumn("ano", F.lit(ANO))\
                                        .withColumn("condicion", F.lit(CONDICION))\
                                        .withColumn("rep", F.lit(REPETICION))

conf_por_sector_final.write.format('parquet')\
    .mode('append')\
    .option('path', 'abfss://silver@covid19arturodatalake.dfs.core.windows.net/conformidad_sector')\
    .save()

# ── Resumen en pantalla ──
print(f"\n{'='*55}")
print(f"  Año: {ANO} | Condición: {CONDICION} | Repetición: {REPETICION}")
print(f"  Registros originales:  {total_original:,}")
print(f"  Total excluidos:       {total_excluidos:,}")
print(f"  Registros FINALES:     {total_final:,}")
print(f"  Porcentaje retenido:   {total_final/total_original*100:.1f}%")
print(f"  Duración:              {duracion_min:.4f} min")
print(f"  Throughput:            {throughput_reg_min:,.0f} reg/min")
print(f"  Bitácora y conformidad guardadas correctamente (Rep {REPETICION})")
print(f"{'='*55}")
```
