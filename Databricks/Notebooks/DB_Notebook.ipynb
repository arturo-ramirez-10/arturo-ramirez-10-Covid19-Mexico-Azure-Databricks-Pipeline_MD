# CREATE SCHEMA

```python
%sql
CREATE SCHEMA COVID19_CATALOG.silver;
```

```python
%sql
CREATE SCHEMA COVID19_CATALOG.gold;
```

```python
%sql
SELECT Ano, Repeticion, timestamp_fin 
FROM parquet.`abfss://silver@covid19arturodatalake.dfs.core.windows.net/bitacora_corridas`
WHERE Ano = '2022' AND Condicion = 'B'
ORDER BY Repeticion
```

```python
%sql
SELECT Ano, Repeticion, timestamp_inicio, timestamp_fin, duracion_min, throughput_reg_min
FROM parquet.`abfss://silver@covid19arturodatalake.dfs.core.windows.net/bitacora_corridas`
ORDER BY Ano, timestamp_fin
```
